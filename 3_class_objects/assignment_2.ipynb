```python
class Person:
    def __init__(self, name=None, age=None):
        self.name = name
        self.age = age
    def show(self):
        print(f"My name is {self.name} and I'm {self.age} years old.")

ahmed = Person("Ahmed")
ahmed.age = 10
ahmed.show()
```

```python
class Circle:
    def __init__(self):
        self.radius = None
    def setRadius(self, radius):
        self.radius = radius
    def getArea(self):
        pi = 3.14
        return pi*(self.radius)**2
    def getCircumference(self):
        pi = 3.14
        return 2*pi*(self.radius)
    
c1 = Circle()
c1.setRadius(10)
print(c1.getArea(), "\n", c1.getCircumference())
```

```python
class Rectangle:
    def __init__(self) -> None:
        self.length = None
        self.breadth = None
    def setDimensions(self, length, breadth):
        self.length = length
        self.breadth = breadth
    def showDimensions(self):
        print(f"Length of rectangle is {self.length} \nBreadth of rectangle is{self.breadth}")
    def getArea(self):
        return f"Area of rectangle is {self.length*self.breadth}"

r1 = Rectangle()
```

```python
class Book:
    def __init__(self):
        self.bookid = 23
        self.title = "Modern Python Book"
        self.price = 1000
    def showbook(self):
        return f"{self.title} is available in Rs\\-{self.price} and it's ID is {self.bookid}"
    
b1 = Book()
print(b1.showbook())
```

```python
class Team:
    def __init__(self) -> None:
        self.list_of_team_member = ["Ali", "Mahad", "Ahmad"]
    def set_member_names(self, list_of_members):
        self.list_of_team_member = list_of_members
    def display_member_names(self):
        return self.list_of_team_member
    
t1 = Team()
print(t1.display_member_names())
t2 = Team()
t2.set_member_names(["Tahir", "Huzaifa"])
print(t2.display_member_names())
```
