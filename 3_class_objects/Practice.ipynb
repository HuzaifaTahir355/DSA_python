Differnce between Class object and Instance object 

```python
class Test:
    x = 5 # Class object Variable / Static Variable
    def __init__(self, b): # Like a constructor, Whenever object is created this function will run autometically.
        self.a = 5 # Instance Object Variable
        self.b = b # self.b is Instance Object Variable but b is a local variable

# Creating an Instance object of Test class
t1 = Test(5) # Instance object 1
t2 = Test(10) # Instance object 2
print(t1.a, t1.b)
print(t2.a, t2.b)
print(t1, t2)

```

```python
class Testing:
    x = 10 # static/class object method
    # Instance method
    def __init__(self):
        self.a = 50

    def show(self):
        print(self.a)

    # static method
    @staticmethod
    def create_st_func():
        print("Hello")
    
    # class method
    @classmethod
    def create_cls_func(cls):
        print(cls.x)

t1 = Testing()

t1.show()

t1.create_st_func()
Testing.create_st_func()

Testing.create_cls_func()
t1.create_cls_func()
```
